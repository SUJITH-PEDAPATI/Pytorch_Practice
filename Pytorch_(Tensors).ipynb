## Comparision Operations

## Performance Comparison: CPU vs. GPU

```python
import time

# Define matrix dimensions
matrix_size = 10000

# Create two large tensors on CPU
a_cpu = torch.rand(matrix_size, matrix_size)
b_cpu = torch.rand(matrix_size, matrix_size)

# Perform matrix multiplication on CPU and measure time
start_time = time.time()
result_cpu = torch.matmul(a_cpu, b_cpu)
end_time = time.time()
print(f"Time taken on CPU: {end_time - start_time:.4f} seconds")

# Check if GPU is available and move tensors to GPU
if torch.cuda.is_available():
  a_gpu = a_cpu.to(device)
  b_gpu = b_cpu.to(device)

  # Perform matrix multiplication on GPU and measure time
  start_time = time.time()
  result_gpu = torch.matmul(a_gpu, b_gpu)
  end_time = time.time()
  print(f"Time taken on GPU: {end_time - start_time:.4f} seconds")
else:
  print("GPU is not available, so cannot perform GPU comparison.")
```

```python
# Squeeze: Remove dimensions of size 1
c = torch.randn(1, 3, 1, 4)
print(f"Original tensor c: {c.shape}")

squeezed_c = c.squeeze()
print(f"Squeezed tensor c: {squeezed_c.shape}")
```

```python
# Unsqueeze: Add a dimension of size 1 at a specified position
d = torch.randn(3, 4)
print(f"Original tensor d: {d.shape}")

unsqueeze_d = d.unsqueeze(0) # Add a new dimension at index 0
print(f"Unsqueeze tensor d (dim 0): {unsqueeze_d.shape}")

unsqueeze_d_2 = d.unsqueeze(1) # Add a new dimension at index 1
print(f"Unsqueeze tensor d (dim 1): {unsqueeze_d_2.shape}")
```
