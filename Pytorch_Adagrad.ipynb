## In Case of a Perceptron

Without Auto Grad

```python
import torch
x = torch.tensor(6.7) # Input Label
y = torch.tensor(0.0) # True Label ( binary )

w = torch.tensor(1.0) # weight
b = torch.tensor(0.0) # bias
```

```python
def binary_cross_entropy_loss(prediction,target):
  epsilon = 1e-8 # To Prevent the Value of log(0) and log(1-1)
  prediction = torch.clamp(prediction,epsilon,1 - epsilon)
  return -(target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

```python
z = w*x + b
```

```python
y_pred = torch.sigmoid(z)
```

```python
loss = binary_cross_entropy_loss(y_pred,y)
```

```python
dloss_dy_pred = (-y/y_pred) + ((1-y)/(1-y_pred))
dy_pred_dz = y_pred * (1-y_pred)
dz_dw = x
dz_db = 1
dloss_dw = dz_dw * dy_pred_dz * dloss_dy_pred
dloss_db = dz_db * dy_pred_dz * dloss_dy_pred
```

```python
print('manual Gradient of Loss w.r.t weight', dloss_dw)
print('manual Gradient of Loss w.r.t bias', dloss_db)
```
