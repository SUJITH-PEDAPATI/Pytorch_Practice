```python
import torch
import torch.nn as nn

"""We are inheriting the properties fom the NN module"""
class Model(nn.Module):
  def __init__(self,num_features):

    """ Calling the Parent Class from the Child Class"""
    super().__init__()
    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out
```

```python
import torch
import torch.nn as nn

"""We are inheriting the properties fom the NN module"""
class Model2(nn.Module):
  def __init__(self,num_features):

    """ Calling the Parent Class from the Child Class"""
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features,3), ## 3 - Output of the Hidden Layer
        nn.ReLU(),
        nn.Linear(3,1), ## 3 - Iput of the final Layer and 1 - Output of the Last layer
        nn.Sigmoid()
    )

  def forward(self,features):
    out = self.network(features)
    return out
```

```python
features = torch.rand(10,5)

"""Creating a instance of Model"""
model = Model2(features.shape[1])

model(features) ## Automatically the Forward Function is trigerred
```

```python
print(model.network[0].weight)
print(model.network[2].weight)
```
